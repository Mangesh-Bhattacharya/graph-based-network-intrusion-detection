# Retired — split into two dataset-specific notebooks

This combined notebook has been replaced. It mixed the Kaggle graph dataset
and the CIC-IDS-2017 raw flow logs in one file, and — worth stating plainly —
it had a real bug: the cell that pulled in DDoS flows via
`load_flow_logs(FRIDAY_DDOS_CSV, nrows=200)` printed **"0 DDoS-labeled flows
pulled in"**, because CIC-IDS-2017 files are stored chronologically and the
first 200 rows of the DDoS file are all benign. That meant the "attack" plot
in this notebook never actually contained an attack.

Please use these two notebooks instead:

- **`01_graph_construction_and_visualization_kaggle.ipynb`** — the pre-built
  Kaggle graph dataset (`data/kaggle/`): all three variants (plain/multi/
  aggregated), degree distribution, hub structure, connected components,
  where attacks sit structurally, and feature extraction — all on the
  complete, real 41,073-node graph.
- **`02_graph_construction_and_visualization_ids_2017.ipynb`** — the raw
  CIC-IDS-2017 flow logs (`data/cic-ids-2017/`): builds communication graphs
  from scratch for a benign baseline day plus four real attack types (DDoS,
  PortScan, Web Attack, Infiltration), using a fixed `sample_labeled_flows()`
  helper that correctly finds real attack rows instead of the buggy
  `nrows=200` approach used here.

Both notebooks were executed end-to-end against the real project datasets
with zero errors before delivery. This file is kept only so the git history
and any existing links to it are not broken; it is not meant to be run.
